```python
import pandas as jo

i = jo.read_csv("fastfood.csv")

i = i[
    [
        "cal_fat",
        "total_fat",
        "trans_fat",
        "cholesterol",
        "sodium",
        "total_carb",
        "fiber",
        "sugar",
        "protein",
        "vit_a",
        "vit_c",
        "calcium",
        "calories",
    ]
]
i = i.fillna(0)

# memisahkan data input (x) dan output (y)
x = i.drop("calories", axis=1)
y = i["calories"]


from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

from sklearn.linear_model import LinearRegression

# Memanggil model  linear
model = LinearRegression()

# memasukkan data ke dalam model
model.fit(X = X_train, y = y_train)

y_pred = model.predict(X_test)

from sklearn.metrics import mean_absolute_error, mean_squared_error

import numpy as jo
jo.sqrt(mean_squared_error(y_test, y_pred))


test_residual = y_test - y_pred
import matplotlib.pyplot as plt
import seaborn as sns

sns.scatterplot(x=y_test, y=test_residual)
plt.axhline(y=0, color='green', linestyle='--')

# Membuat grafik histogram dan kurva kepadatan residual
plt.figure(figsize=(6, 4))
sns.histplot(test_residual, kde=True, bins=25, color='blue')

# Memberi label teks
plt.xlabel('Nilai Kesalahan / Residual (y_test - y_pred)')
plt.ylabel('Jumlah Frekuensi Data')
plt.title('Grafik Distribusi Normal Kesalahan AI')
plt.show()



```
